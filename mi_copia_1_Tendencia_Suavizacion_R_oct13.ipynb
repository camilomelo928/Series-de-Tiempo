## **1. Método de Descomposición**

El método de descomposición en series de tiempo, es una técnica analítica utilizada para entender la estructura y los patrones presentes en una serie temporal, la cual ayuda a separar la serie en componentes fundamentales, lo que facilita el análisis y la comprensión de las tendencias, estacionalidades y patrones cíclicos presentes en los datos.

La descomposición generalmente se realiza mediante un enfoque aditivo o multiplicativo, la cual incluye las siguientes componentes:

   - **Tendencia ($T$)**: Representa la dirección general de los datos a lo largo del tiempo, identificando si la serie muestra una tendencia creciente o decreciente.

   - **Estacionalidad ($S$)**: Refleja patrones estacionales o cíclicos que se repiten a intervalos fijos de tiempo, como estacionalidades mensuales o anuales.

   - **Residuo ($R$)**: Es la parte restante de la serie que no puede explicarse mediante la tendencia y la estacionalidad. Incluye variaciones aleatorias o errores.

---

1. **Descomposición Aditiva:**

La serie temporal original se puede expresar como la suma de estos componentes: 
$$X_t = T_t + S_t + R_t.$$

2. **Descomposición Multiplicativa:**

En lugar de sumar los componentes, se multiplican. La serie temporal original se expresa como: 
$$X_t = T_t \cdot S_t \cdot R_t.$$

## **2. Regresión Lineal en Series de Tiempo**

La **Regresión Lineal** para Series de Tiempo es un método que busca modelar la relación lineal entre una variable dependiente y el tiempo. Utiliza datos históricos para encontrar una línea de mejor ajuste que represente la tendencia a lo largo del tiempo. Este modelo permite prever futuros valores basados en la relación identificada. Sin embargo, la regresión lineal asume que la relación entre la variable y el tiempo es constante, lo que puede ser limitante para capturar patrones más complejos en series temporales. Otros métodos, como modelos autoregresivos o de suavizado exponencial, son a menudo preferidos para abordar la dinámica temporal más efectivamente.

```r
# Definición de los Tiempos #
t <- as.vector(time(jj))
t2 <- t^2
```

```r
# Modelo Lineal names(mod.lin) con resumen #
mod.lin <- lm(jj~t)
summary(mod.lin)
```

```r
# Modelo Cuadratico #
mod.cuad <- lm(jj~t+t2);
summary(mod.cuad)
```

```r
# Modelo Log-Lineal #
mod.log.lin = lm(log(jj)~t) 
summary(mod.log.lin)
```

```r
# Modelo Exponencial #
Ds = data.frame(jj,t)
beta0 = mod.log.lin$coefficient[1]
beta1 = mod.log.lin$coefficient[2]
mod.exp = nls(jj~exp(beta0+beta1*t),data=Ds,
start=list(beta0=beta0, beta1=beta1)); summary(mod.exp)
```

```r
# DataFrame
df <- data.frame(
  Tiempo = t,
  Ganancias = as.numeric(jj),
  Lineal = predict(mod.lin),
  Cuadratico = as.numeric(mod.cuad$coefficients[1]) +
               as.numeric(mod.cuad$coefficients[2]) * t +
               as.numeric(mod.cuad$coefficients[3]) * t2,
  Exponencial = exp(mod.log.lin$coefficients[1] +
                    mod.log.lin$coefficients[2] * t),
  Exponencial_Iterado = exp(summary(mod.exp)$coefficients[1] +
                            summary(mod.exp)$coefficients[2] * t)
)

# Gráfico interactivo
plot_ly(df) %>%
  add_trace(
    x = ~Tiempo,
    y = ~Ganancias,
    type = "scatter",
    mode = "lines+markers",
    name = "Serie",
    line = list(color = "black", width = 2),
    marker = list(size = 5),
    hovertemplate = "<b>Año:</b> %{x:.2f}<br><b>EPS:</b> %{y:.2f}<extra></extra>"
  ) %>%
  add_trace(
    x = ~Tiempo,
    y = ~Lineal,
    type = "scatter",
    mode = "lines",
    name = "Lineal",
    line = list(color = "#E41A1C", width = 3)
  ) %>%
  add_trace(
    x = ~Tiempo,
    y = ~Cuadratico,
    type = "scatter",
    mode = "lines",
    name = "Cuadrático",
    line = list(color = "#377EB8", width = 3)
  ) %>%
  add_trace(
    x = ~Tiempo,
    y = ~Exponencial,
    type = "scatter",
    mode = "lines",
    name = "Exponencial",
    line = list(color = "#4DAF4A", width = 3, dash = "dot")
  ) %>%
  add_trace(
    x = ~Tiempo,
    y = ~Exponencial_Iterado,
    type = "scatter",
    mode = "lines",
    name = "Exponencial Iterado",
    line = list(color = "#FF8C00", width = 3, dash = "dash")
  ) %>%
  layout(
    title = list(
      text = "<b>Johnson & Johnson: Serie con Tendencias Estimadas</b>",
      x = 0.5, y=0.98
    ),
    xaxis = list(
      title = "Tiempo",
      showgrid = TRUE,
      gridcolor = "lightgray",
      rangeslider = list(visible = FALSE)
    ),
    yaxis = list(
      title = "Quarterly Earnings per Share",
      range = c(-2, 17),
      showgrid = TRUE,
      gridcolor = "lightgray",
      zeroline = FALSE
    ),
    legend = list(
  orientation = "v",
  x = 1.02,
  y = 1,
  xanchor = "left",
  yanchor = "top",
  bgcolor = "rgba(255,255,255,0.8)",
  bordercolor = "lightgray",
  borderwidth = 1,
  font = list(size = 12)
),
    hovermode = "x unified",
    plot_bgcolor = "white",
    height = 800,
    paper_bgcolor = "white"
  ) %>%
  config(
    displaylogo = FALSE,
    scrollZoom = TRUE,
    toImageButtonOptions = list(format = "png")
  )
```

```r
# Criterios de Información AIC y BIC #

AIC.tot = c(AIC(mod.lin),
            AIC(mod.cuad),
            AIC(mod.exp))
BIC.tot = c(AIC(mod.lin,k=log(length(jj))),
            AIC(mod.cuad,k=log(length(jj))),
            AIC(mod.exp,k=log(length(jj))))
```

```r
# Comparación de valores de AIC y BIC para cada modelo #

library(knitr)

medidas <- rbind(AIC.tot, BIC.tot)
colnames(medidas) <- c("Lineal", "Cuadrático", "Exp. Iterado")
rownames(medidas) <- c("AIC", "BIC")

kable(
  medidas,
  digits = 2,
  align = "c",
  caption = "Comparación de los criterios de información"
)
```

## **3. Media Móvil (Filtros)**
Es un método que calcula el promedio de un conjunto de observaciones sucesivas a lo largo del tiempo. Consiste en tomar un número fijo de puntos de datos (la ventana de la media móvil) y calcular el promedio en cada paso, desplazando la ventana a medida que avanzan los datos. Esto suaviza la serie temporal, eliminando el ruido aleatorio y resaltando las tendencias subyacentes. La elección del tamaño de la ventana afecta la sensibilidad del método. Una ventana más grande suaviza la serie más agresivamente, mientras que una ventana más pequeña captura variaciones más rápidas. La media móvil es una técnica simple pero útil para identificar patrones y tendencias en los datos temporales.

## **4. Suavización Exponencial Simple (SES)**

Pondera todos los datos pasados con ponderaciones que disminuyen exponencialmente al ir hacia el pasado. Es decir, normalmente los datos más reciente tienen una mayor ponderaci ́on. La ponderación de esta forma supera en gran medida las limitaciones de promedios m ́oviles o m ́etodos de cambio de porcentaje.
Este método, cuyo resultado es una previsión plana, recta, es el mejor para datos volátiles sin tendencia ni estacionalidad.
La constante funciona como un factor de ponderación y su variación se hace de acuerdo a nuestra necesidad de darle más peso a datos recientes (α más elevado) o a datos pasados (α más bajo). Si α = 1, nuestro pronóstico del próximo periodo será exactamente igual al del periodo actual.

</div>

```r
library(forecast)
library(ggplot2)
library(plotly)
options(warn = -1)
#=========================================================
# Ajuste con Suavización Exponencial
#=========================================================
options(warn = -1)
fit_close <- ses(
  close,
  h = 5,
  initial = "simple",
  alpha = 0.3
)

#=========================================================
# Serie observada y ajustada
#=========================================================
df_obs <- data.frame(
  Fecha = as.Date(time(close)),
  Observado = as.numeric(close),
  Ajustado = as.numeric(fitted(fit_close))
)

#=========================================================
# Pronósticos
#=========================================================
ultima_fecha <- max(df_obs$Fecha)

df_pred <- data.frame(
  Fecha = seq(
    from = ultima_fecha + 1,
    by = "day",
    length.out = length(fit_close$mean)
  ),
  Pronostico = as.numeric(fit_close$mean),
  LI95 = fit_close$lower[,2],
  LS95 = fit_close$upper[,2],
  LI80 = fit_close$lower[,1],
  LS80 = fit_close$upper[,1]
)

#=========================================================
# Gráfico
#=========================================================
g <- ggplot() +

  ## Intervalo 95%
  geom_ribbon(
    data = df_pred,
    aes(
      x = Fecha,
      ymin = LI95,
      ymax = LS95,
      fill = "95%"
    ),
    alpha = 0.20
  ) +

  ## Intervalo 80%
  geom_ribbon(
    data = df_pred,
    aes(
      x = Fecha,
      ymin = LI80,
      ymax = LS80,
      fill = "80%"
    ),
    alpha = 0.35
  ) +

  ## Serie observada
  geom_line(
    data = df_obs,
    aes(
      Fecha,
      Observado,
      colour = "Serie observada"
    ),
    linewidth = 0.9
  ) +

  ## Serie ajustada
  geom_line(
    data = df_obs,
    aes(
      Fecha,
      Ajustado,
      colour = "Suavización exponencial"
    ),
    linewidth = 1
  ) +

  ## Pronóstico
  geom_line(
    data = df_pred,
    aes(
      Fecha,
      Pronostico,
      colour = "Pronóstico"
    ),
    linewidth = 1
  ) +

  geom_point(
    data = df_pred,
    aes(
      Fecha,
      Pronostico
    ),
    colour = "#d62728",
    size = 1
  ) +

  ## Inicio del horizonte de pronóstico
  geom_vline(
    xintercept = ultima_fecha,
    linetype = "dashed",
    colour = "grey40"
  ) +

  scale_colour_manual(
    values = c(
      "Serie observada" = "black",
      "Suavización exponencial" = "#1f77b4",
      "Pronóstico" = "#d62728"
    )
  ) +

  scale_fill_manual(
    values = c(
      "80%" = "#74a9cf",
      "95%" = "#c6dbef"
    )
  ) +

  scale_x_date(
    date_breaks = "2 months",
    date_labels = "%b\n%Y"
  ) +

  labs(
    title = "Pronóstico mediante Suavización Exponencial Simple",
    subtitle = "Precio de cierre (Microsoft)",
    x = "Fecha",
    y = "Close",
    colour = "",
    fill = "Intervalo\nPredicción"
  ) +

  theme_minimal(base_size = 14) +

  theme(
    plot.title = element_text(face = "bold", hjust = 0.5),
    plot.subtitle = element_text(hjust = 0.5),
    legend.position = "right",
    panel.grid.minor = element_blank(),
    panel.grid.major = element_line(colour = "grey85")
  )

#=========================================================
# Plotly
#=========================================================
ggplotly(
  g,
  tooltip = c("x","y","colour","fill")
) %>%
layout(
  hovermode = "x unified",
  plot_bgcolor = "white",
  paper_bgcolor = "white",
  height=500,
  legend = list(
    x = 1.02,
    y = 1,
    xanchor = "left"
  ),
  xaxis = list(
    type = "date",
    rangeslider = list(visible = FALSE)
  )
) %>%
config(
  displaylogo = FALSE,
  scrollZoom = TRUE,
  toImageButtonOptions = list(
    format = "png",
    filename = "Pronostico_SES_Microsoft"
  )
)
```

## **5. Suavización Holt-Winters**
El método de Holt-Winters es una técnica de suavizado exponencial triple utilizada para modelar y predecir series de tiempo con patrones de tendencia y estacionalidad. Este enfoque toma en cuenta tres componentes principales: nivel, tendencia y estacionalidad. Utiliza algoritmos de suavizado exponencial para actualizar y pronosticar estos componentes en cada paso de tiempo. El método de Holt-Winters es útil para series temporales que exhiben tanto tendencias de crecimiento o declive como patrones estacionales regulares. Ayuda a capturar y prever estos patrones de manera más precisa que los métodos de suavizado exponencial simple o doble.

</div>

```r
# Evaluar capacidad predictiva #
pred_vals <- as.numeric(pred_hw[,1])  # valores centrales predichos
real_vals <- as.numeric(test[1:4])    # los 4 valores reales
```

```r
# Métricas de error 
mae <- mean(abs(pred_vals - real_vals)) 
rmse <- sqrt(mean((pred_vals - real_vals)^2)) 
mape <- mean(abs((pred_vals - real_vals) / real_vals)) * 100
```

```r
library(knitr)

# Crear un dataframe con las métricas organizadas
df_metricas <- data.frame(
  Métrica = c("MAE", "RMSE", "MAPE"),
  Descripción = c("Error Absoluto Medio", 
                  "Raíz del Error Cuadrático Medio", 
                  "Error Porcentual Absoluto Medio"),
  Valor = c(round(mae, 3), round(rmse, 3), paste0(round(mape, 2), "%"))
)

# Imprimir la tabla formateada
kable(
  df_metricas, 
  align = c("l", "l", "r"),
  caption = "Métricas de Precisión del Pronóstico"
)
```
